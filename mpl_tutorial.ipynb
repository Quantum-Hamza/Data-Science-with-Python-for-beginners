```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline

plt.style.use('https://github.com/dhaitz/matplotlib-stylesheets/raw/master/pitayasmoothie-dark.mplstyle')
```

```python
df = pd.read_csv("bitcoin_dataset.csv", index_col=0, parse_dates=True)
df
```

```python
plt.plot(df.index, df["btc_market_price"])
plt.xlabel('Date')
plt.ylabel('Market Price')
plt.title('BTC Market Price')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(df.index, df["btc_market_price"])
ax.set_xlabel('Date')
ax.set_ylabel('Market Price')
ax.set_title('BTC Market Price')
plt.show()
```

## Basic line plots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sine Wave')
plt.grid(True)
plt.axhline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.axvline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.plot(x, y3, label='sin(x) + cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trig Functions')
plt.grid(True, alpha=0.75)
plt.axhline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.axvline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.legend()
plt.show()
```

## Customization

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure(figsize=(12, 6))
plt.plot(x, y1, linestyle='--', label='sin(x)')
plt.plot(x, y2, linestyle=':', label='cos(x)')
plt.plot(x, y3, linewidth=3, label='sin(x) * cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trig Functions')
plt.grid(True, alpha=0.75)
plt.axhline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.axvline(0, color='lightblue', linewidth=0.5) # IMPORTANT TO SPECIFY
plt.legend()
plt.show()
```
