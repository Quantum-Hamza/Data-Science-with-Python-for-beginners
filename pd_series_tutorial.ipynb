```python
import pandas as pd
import numpy as np
```

```python
g7_pop = pd.Series([34, 45, 45, 56, 234, 56, 34])
g7_pop
```

```python
print(g7_pop[0])
```

**It is better than a python list because it also provides index customization:**

```python
g7_pop.index = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
print(g7_pop)
```

**It is also better than a dictionary because it is ordered**

```python
#You can write both elements and indexes at once
pd.Series([34, 45, 45, 56, 234, 56, 34],
          index=['a', 'b', 'c', 'd', 'e', 'f', 'g'],
         name="This is a Series example")
```

```python
pd.Series(g7_pop, index=['a', 'b', 'c', 'd', 'z'])
```

## Indexing

```python
print(g7_pop[['a', 'b']])
```

```python
print(g7_pop.iloc[[0, 1]])
```

**CAREFUL: UPPER LIMIT IS *INCLUDED* IN PANDAS**

```python
g7_pop['a': 'c']
```

## Boolean Series

```python
g7_pop > 50
```

```python
g7_pop[(g7_pop > 50) | (g7_pop < 40)]
```

## Operations and methods

```python
g7_pop * 100
```

## Modifying series

```python
g7_pop['a'] = 47
g7_pop.iloc[1] = 48
g7_pop
```

```python
g7_pop[g7_pop < 50] = 0
g7_pop
```
